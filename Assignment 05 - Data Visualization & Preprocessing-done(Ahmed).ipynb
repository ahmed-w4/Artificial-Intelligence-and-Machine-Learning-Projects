```python
# Remove outliers if needed
Q1 = df['avg_glucose_level'].quantile(0.25)
Q3 = df['avg_glucose_level'].quantile(0.75)

IQR = Q3-Q1
Q1, Q3, IQR
lower_bound = Q1 - (1.5*IQR)
upper_bound = Q3 + (1.5*IQR)


outliers = df[(df['avg_glucose_level']<lower_bound) | 
              (df['avg_glucose_level']>upper_bound)]

print(f'Number of outliers are {len(outliers)}')

df_clean = df[(df['avg_glucose_level'] >= lower_bound) & 
              (df['avg_glucose_level'] <= upper_bound)]

df_2 = df.copy()
df_2 = df_2.drop(outliers.index, axis=0)

df_2
```
