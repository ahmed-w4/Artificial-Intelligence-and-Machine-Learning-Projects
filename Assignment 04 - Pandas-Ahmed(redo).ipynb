## Task 2: Handling Missing Values

The dataset contains missing values in the `Salary` and `Bonus` columns.

**Requirements:**
- Fill the missing `Salary` values with the median salary of the entire company
- Fill the missing `Bonus` values with 0

```python

salary_median = employees['Salary'].median()
salary_median

employees['Salary'].fillna(salary_median, inplace=True)

employees['Bonus'].fillna(0, inplace=True)

employees
```

## Task 3: Data Cleaning

The `Employee_Name` column is in the format "FirstName LastName".

**Requirements:**
- Split this column into two new columns: `First_Name` and `Last_Name`
- Delete the original `Employee_Name` column

```python
# import pandas as pd

# employees = pd.read_csv('C:/Users/Ahmed Wael/.vscode/company_data.csv')

employees[['First_Name', 'Last_Name']] = employees['Employee_Name'].str.split(' ', expand=True)

employees.drop('Employee_Name', axis=1, inplace=True)

employees
```

## Task 5: Performance Mapping

Create a new column called `Performance_Level` based on the `Evaluation_Score` (1-100).

**Requirements:**
- Score > 80: "High"
- Score 50-80: "Medium"
- Score < 50: "Low"

```python
import pandas as pd
employees = pd.read_csv('C:/Users/Ahmed Wael/.vscode/company_data.csv')

def get_performance(score):
    if score >80:
        return 'High'
    elif score >=50:
        return 'Medium'
    else:
        return 'Low'

employees['Performance_Level']= employees['Evaluation_Score'].apply(get_performance)

employees
```

## Task 6: Department Statistics

Calculate statistics for each Department.

**Requirements:**
- Calculate the total salary expense for each department
- Calculate the average evaluation score for each department

```python
import pandas as pd
employees = pd.read_csv('C:/Users/Ahmed Wael/.vscode/company_data.csv')

total_salary = employees.groupby('Department')['Salary'].sum().round(2)
average_evaluation = employees.groupby('Department')['Evaluation_Score'].mean().round(2)
print('Total salary for each department\n', total_salary)
print()
print('Average Evaluation Scores for each department\n', average_evaluation)
```

## Task 8: Salary Update
